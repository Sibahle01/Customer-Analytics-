```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.cluster import KMeans
import xgboost as xgb
from sklearn.metrics import classification_report, confusion_matrix
import os
from datetime import datetime, timedelta


os.makedirs('data/processed', exist_ok=True)
os.makedirs('visualizations', exist_ok=True)

# 1. Load the CSV files directly
customers_df = pd.read_csv('/content/drive/MyDrive/customers.csv')
transactions_df = pd.read_csv('/content/drive/MyDrive/transactions.csv')
creditcard_df = pd.read_csv('/content/drive/MyDrive/creditcard.csv')


print("Original customer data shape:", customers_df.shape)


customers_df['income_category'] = customers_df['income_category'].fillna('Unknown')


customers_df = customers_df.drop_duplicates(subset=['customer_id'])

customers_df['age'] = pd.to_numeric(customers_df['age'], errors='coerce')
customers_df['tenure_months'] = pd.to_numeric(customers_df['tenure_months'], errors='coerce')
customers_df['credit_score'] = pd.to_numeric(customers_df['credit_score'], errors='coerce')  # Note: Fixed typo in column name


customers_df['age'] = customers_df['age'].fillna(customers_df['age'].median())
customers_df['tenure_months'] = customers_df['tenure_months'].fillna(customers_df['tenure_months'].median())
customers_df['credit_score'] = customers_df['credit_score'].fillna(customers_df['credit_score'].median())

# Rename 'credic_score' to 'credit_score' to fix typo
customers_df = customers_df.rename(columns={'credic_score': 'credit_score'})

print("Cleaned customer data shape:", customers_df.shape)


print("Original transactions data shape:", transactions_df.shape)


transactions_df['transaction_datetime'] = pd.to_datetime(transactions_df['transaction_datetime'])


transactions_df = transactions_df.rename(columns={'is_fraud': 'is_flagged'})


transactions_df['hour'] = transactions_df['transaction_datetime'].dt.hour
transactions_df['day'] = transactions_df['transaction_datetime'].dt.dayofweek
transactions_df['month'] = transactions_df['transaction_datetime'].dt.month


transactions_df.loc[transactions_df['amount'] < 10, 'amount'] = transactions_df['amount'] * 18.5


transactions_df = transactions_df.merge(
    customers_df[['customer_id', 'province']],
    on='customer_id',
    how='left'
)


print("Original creditcard data shape:", creditcard_df.shape)


creditcard_df = creditcard_df.rename(columns={
    'Class': 'is_flagged',
    'Amount': 'amount',
    'Time': 'time_seconds'
})

print("Cleaned transactions data shape:", transactions_df.shape)


transaction_analysis = transactions_df.groupby('customer_id').agg({
    'transaction_id': 'count',
    'amount': ['sum', 'mean'],
    'is_flagged': 'sum'
}).reset_index()


transaction_analysis.columns = ['customer_id', 'tx_count', 'total_spent', 'avg_transaction', 'fraud_count']


analysis_df = transaction_analysis.merge(
    customers_df,
    on='customer_id',
    how='left'
)


customer_tx_dates = transactions_df.groupby('customer_id')['transaction_datetime'].agg(['min', 'max', 'count'])
customer_tx_dates['date_range'] = (customer_tx_dates['max'] - customer_tx_dates['min']).dt.days
customer_tx_dates['tx_frequency'] = customer_tx_dates['date_range'] / customer_tx_dates['count']
customer_tx_dates = customer_tx_dates.replace([np.inf, -np.inf], np.nan)
customer_tx_dates['tx_frequency'] = customer_tx_dates['tx_frequency'].fillna(0)


analysis_df = analysis_df.merge(
    customer_tx_dates[['tx_frequency']],
    left_on='customer_id',
    right_index=True,
    how='left'
)

# Save processed data for PowerBI
customers_df.to_csv('data/processed/customers_processed.csv', index=False)
transactions_df.to_csv('data/processed/transactions_processed.csv', index=False)
creditcard_df.to_csv('data/processed/creditcard_processed.csv', index=False)
analysis_df.to_csv('data/processed/analysis_data.csv', index=False)

print("Analysis dataframe shape:", analysis_df.shape)
print("Data preparation complete!")
```

```python
# Prepare features
print("Building fraud detection model using creditcard dataset...")
fraud_features = [col for col in creditcard_df.columns if col.startswith('V')] + ['amount']
X = creditcard_df[fraud_features]
y = creditcard_df['is_flagged']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train XGBoost model
fraud_model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
fraud_model.fit(X_train, y_train)

y_pred = fraud_model.predict(X_test)
print("Fraud Detection Model Performance:")
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))


feature_importance = pd.DataFrame({
    'feature': X.columns,
    'importance': fraud_model.feature_importances_
}).sort_values('importance', ascending=False)

# Save feature importance for PowerBI
feature_importance.to_csv('data/processed/feature_importance.csv', index=False)


print("Applying fraud model to transaction dataset...")

if set([col for col in creditcard_df.columns if col.startswith('V')]).issubset(set(transactions_df.columns)):
    # If transaction data has V1-V28 features, apply model directly
    transaction_features = fraud_features
    X_transactions = transactions_df[transaction_features]
    # Predict fraud probability
    fraud_probs = fraud_model.predict_proba(X_transactions)[:, 1]
    transactions_df['fraud_probability'] = fraud_probs
else:

    print("V1-V28 features not found in transaction data. Training a simpler model.")

    # Create a new model based on common features between datasets
    common_features = ['amount']

    # Add transaction-specific features
    tx_features = ['hour', 'day']
    categorical_features = ['transaction_type', 'merchant_category']

    # Prepare X and y for the transaction-specific model
    X_tx = transactions_df[common_features + tx_features]
    y_tx = transactions_df['is_flagged']


    X_tx = pd.get_dummies(transactions_df, columns=categorical_features, drop_first=True)

    # Keep only numeric columns for modeling
    numeric_cols = X_tx.select_dtypes(include=[np.number]).columns
    X_tx = X_tx[numeric_cols]


    X_tx_train, X_tx_test, y_tx_train, y_tx_test = train_test_split(X_tx, y_tx, test_size=0.3, random_state=42)


    tx_fraud_model = RandomForestClassifier(n_estimators=100, random_state=42)
    tx_fraud_model.fit(X_tx_train, y_tx_train)


    y_tx_pred = tx_fraud_model.predict(X_tx_test)
    print("Transaction-Specific Fraud Model Performance:")
    print(classification_report(y_tx_test, y_tx_pred))


    fraud_probs = tx_fraud_model.predict_proba(X_tx)[:, 1]
    transactions_df['fraud_probability'] = fraud_probs


    tx_feature_importance = pd.DataFrame({
        'feature': X_tx.columns,
        'importance': tx_fraud_model.feature_importances_
    }).sort_values('importance', ascending=False)


    tx_feature_importance.to_csv('data/processed/tx_feature_importance.csv', index=False)


transactions_df.to_csv('data/processed/transactions_with_fraud_probs.csv', index=False)

feature_importance.to_csv('data/processed/feature_importance.csv', index=False)


print("Performing customer segmentation...")

segment_features = ['age', 'tx_count', 'total_spent', 'avg_transaction', 'credit_score']
X_segment = analysis_df[segment_features].copy()

# Handle missing values
for col in X_segment.columns:
    X_segment[col] = X_segment[col].fillna(X_segment[col].median())

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_segment)

# Determine optimal number of clusters
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Apply KMeans with 4 clusters
kmeans = KMeans(n_clusters=4, random_state=42)
analysis_df['segment'] = kmeans.fit_predict(X_scaled)


segment_profile = analysis_df.groupby('segment').agg({
    'age': 'mean',
    'tx_count': 'mean',
    'total_spent': 'mean',
    'avg_transaction': 'mean',
    'customer_id': 'count'
}).reset_index()

print("Customer Segment Profiles:")
print(segment_profile)


segment_names = {
    0: "High-Value Loyalists",
    1: "Young Professionals",
    2: "Budget Savers",
    3: "Occasional Users"
}

analysis_df['segment_name'] = analysis_df['segment'].map(segment_names)


analysis_df.to_csv('data/processed/segmented_customers.csv', index=False)
segment_profile.to_csv('data/processed/segment_profiles.csv', index=False)

print("Data analysis and modeling complete!")
```

```python


print("Creating visualizations...")

# 1. Fraud Detection Visualizations
plt.figure(figsize=(18, 12))

# Fraud by hour of day
plt.subplot(2, 2, 1)
hourly_data = transactions_df.groupby('hour').agg({
    'transaction_id': 'count',
    'is_flagged': 'sum'
}).reset_index()
hourly_data['fraud_rate'] = hourly_data['is_flagged'] / hourly_data['transaction_id'] * 100

sns.lineplot(x='hour', y='fraud_rate', data=hourly_data)
plt.title('Fraud Rate by Hour of Day')
plt.xlabel('Hour')
plt.ylabel('Fraud Rate (%)')

# Fraud by transaction amount
plt.subplot(2, 2, 2)
sns.boxplot(x='is_flagged', y='amount', data=transactions_df)
plt.title('Transaction Amount by Fraud Status')
plt.xlabel('Fraudulent (1) vs. Legitimate (0)')
plt.ylabel('Amount (ZAR)')

# Feature importance for fraud detection
plt.subplot(2, 2, 3)
top_features = feature_importance.head(10)
sns.barplot(x='importance', y='feature', data=top_features)
plt.title('Top 10 Features for Fraud Detection')
plt.xlabel('Importance')
plt.ylabel('Feature')

# Fraud by province
plt.subplot(2, 2, 4)
province_fraud = transactions_df.groupby('province').agg({
    'transaction_id': 'count',
    'is_flagged': 'sum'
}).reset_index()
province_fraud['fraud_rate'] = province_fraud['is_flagged'] / province_fraud['transaction_id'] * 100

sns.barplot(x='province', y='fraud_rate', data=province_fraud)
plt.title('Fraud Rate by Province')
plt.xlabel('Province')
plt.ylabel('Fraud Rate (%)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig('visualizations/fraud_analysis.png', dpi=300)
plt.close()

# 2. Customer Segmentation Visualizations
plt.figure(figsize=(18, 12))

# Segment distribution
plt.subplot(2, 2, 1)
segment_counts = analysis_df['segment_name'].value_counts()
plt.pie(segment_counts, labels=segment_counts.index, autopct='%1.1f%%')
plt.title('Customer Segment Distribution')

# Age distribution by segment
plt.subplot(2, 2, 2)
sns.boxplot(x='segment_name', y='age', data=analysis_df)
plt.title('Age Distribution by Customer Segment')
plt.xlabel('Segment')
plt.ylabel('Age')
plt.xticks(rotation=45)

# Spending by segment
plt.subplot(2, 2, 3)
sns.barplot(x='segment_name', y='total_spent', data=analysis_df)
plt.title('Average Total Spending by Segment')
plt.xlabel('Segment')
plt.ylabel('Total Spent (ZAR)')
plt.xticks(rotation=45)

# Transaction frequency by segment
plt.subplot(2, 2, 4)
sns.barplot(x='segment_name', y='tx_count', data=analysis_df)
plt.title('Average Transaction Count by Segment')
plt.xlabel('Segment')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig('visualizations/customer_segmentation.png', dpi=300)
plt.close()


transactions_df['month_year'] = transactions_df['transaction_datetime'].dt.to_period('M')
monthly_trends = transactions_df.groupby('month_year').agg({
    'transaction_id': 'count',
    'amount': 'sum',
    'is_flagged': 'sum'
}).reset_index()
monthly_trends['month_year'] = monthly_trends['month_year'].astype(str)
monthly_trends.to_csv('data/processed/monthly_trends.csv', index=False)

# Transaction types analysis
tx_type_analysis = transactions_df.groupby(['transaction_type', 'is_flagged']).agg({
    'transaction_id': 'count',
    'amount': 'sum'
}).reset_index()
tx_type_analysis.to_csv('data/processed/transaction_types.csv', index=False)

# Merchant category analysis
merchant_analysis = transactions_df.groupby(['merchant_category', 'is_flagged']).agg({
    'transaction_id': 'count',
    'amount': 'sum'
}).reset_index()
merchant_analysis.to_csv('data/processed/merchant_categories.csv', index=False)

print("Visualizations created and saved to the 'visualizations' folder!")
print("Additional analysis data saved for PowerBI!")
```
